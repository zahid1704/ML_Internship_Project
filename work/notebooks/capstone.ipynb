## Demo outline (5 minutes)

**1. The question (45 seconds).** Content teams can only review a few dozen pages a week. Which pages should they look at first when a page is seen often but clicked rarely? A fair answer has to compare each page with others at the same search position.

**2. The method (1 minute 30 seconds).** I used March 2026 search signals from the FlyRank internship data to rank pages, and checked the ranking against April 2026. The baseline is a readable rule: rank pages by the clicks they would have earned at their position tier's typical click-through rate, minus the clicks they actually got. I compared it with a random forest, holding out whole clients so the test used sites the model had never seen, and I audited for leakage.

**3. One chart (45 seconds).** Median click-through rate by position tier (positions 1-3, 4-10, 11-20, 21+). It shows why position has to be accounted for: pages near the top get far more clicks, so comparing across positions would be unfair.

**4. One honest result (1 minute).** On 12 held-out clients, the random forest reached an AUC of 0.799 against 0.774 for the simple rule. That is a small, directional edge, and my first version overstated it until I found a leak in how the typical CTRs were computed. A deliberately leaked feature pushed AUC to 0.989, which shows why the audit mattered.

**5. One recommendation (1 minute).** Use the simple rule to order the review queue, with the model as a second opinion, and have a person check each top page (tracking, search results page, query intent) before editing anything. The queue says where to look first, not why a CTR is low or that an edit will raise clicks.
